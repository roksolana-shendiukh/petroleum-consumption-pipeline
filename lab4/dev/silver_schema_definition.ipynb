```python
from pyspark.sql.types import *

consumption_schema = StructType([
    StructField("consumption_sk", StringType(), False),
    StructField("series_bk", StringType(), False),
    StructField("duoarea_bk", StringType(), False),
    StructField("period_bk", DateType(), False),
    StructField("product_bk", StringType(), True),
    StructField("units", StringType(), True),
    StructField("consumption_value", DecimalType(10,3), True),
    StructField("_source_system", StringType(), False),
    StructField("_ingested_at", TimestampType(), False),
])

prices_schema = StructType([
    StructField("price_sk", StringType(), False),
    StructField("series_bk", StringType(), False),
    StructField("product_name", StringType(), True),
    StructField("units", StringType(), True),
    StructField("price", DecimalType(10,3), True),
    StructField("effective_from", DateType(), False),
    StructField("effective_to", DateType(), True),  
    StructField("is_current", BooleanType(), False),
    StructField("_source_system", StringType(), False),
    StructField("_ingested_at", TimestampType(), False),
])
```

```python
spark.createDataFrame([], consumption_schema).write.format("delta") \
    .mode("errorifexists") \
    .saveAsTable("<login>_silver.petroleum_consumption")

spark.createDataFrame([], prices_schema).write.format("delta") \
    .mode("errorifexists") \
    .saveAsTable("<login>_silver.petroleum_prices")
```
